```python
# NDMU Student Grade Report System — Google Colab
# Searchable student list, automatic individual reports, weighted GWA, and Top 10.
# The uploaded source file is read only; this application does not overwrite it.

import io, re, base64
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
from html import escape
from google.colab import files

GREEN = "#14532D"
DARK_GREEN = "#0B2F24"
GOLD = "#F4D16B"
PALE_GOLD = "#FFF8DF"
LOGO_B64 = ""

def clean_text(value):
    if pd.isna(value):
        return ""
    return str(value).strip()

def norm(value):
    return re.sub(r"[^a-z0-9]+", " ", clean_text(value).lower()).strip()

def find_column(columns, choices):
    normalized = {norm(c): c for c in columns}
    for choice in choices:
        if norm(choice) in normalized:
            return normalized[norm(choice)]
    return None

def parse_repeated_sections(raw):
    raw = raw.copy().reset_index(drop=True)
    header_rows = []
    for i in range(len(raw)):
        values = [norm(v) for v in raw.iloc[i].tolist()]
        if "subject code" in values and "grade" in values and "units" in values:
            header_rows.append(i)

    records = []
    if header_rows:
        for header_i in header_rows:
            if header_i == 0:
                continue
            info = raw.iloc[header_i - 1]
            student_name = clean_text(info.iloc[0])
            course = clean_text(info.iloc[1]) if len(info) > 1 else ""
            year = clean_text(info.iloc[2]) if len(info) > 2 else ""
            section_headers = [norm(v) for v in raw.iloc[header_i].tolist()]

            def col_index(*names):
                for name in names:
                    if name in section_headers:
                        return section_headers.index(name)
                return None

            code_i = col_index("subject code", "subject")
            desc_i = col_index("subject description", "description")
            units_i = col_index("units", "unit")
            teacher_i = col_index("teacher", "instructor", "professor")
            grade_i = col_index("grade", "final grade")

            for j in range(header_i + 1, len(raw)):
                row = raw.iloc[j]
                if all(clean_text(v) == "" for v in row.tolist()):
                    break
                if "subject code" in [norm(v) for v in row.tolist()]:
                    break
                code = clean_text(row.iloc[code_i]) if code_i is not None else ""
                desc = clean_text(row.iloc[desc_i]) if desc_i is not None else ""
                teacher = clean_text(row.iloc[teacher_i]) if teacher_i is not None else ""
                units = pd.to_numeric(row.iloc[units_i], errors="coerce") if units_i is not None else float("nan")
                grade = pd.to_numeric(row.iloc[grade_i], errors="coerce") if grade_i is not None else float("nan")
                if not code or code.lower() in ("nan", "subject code"):
                    continue
                if pd.isna(units) or pd.isna(grade) or units <= 0:
                    continue
                records.append({
                    "student_name": student_name, "course": course, "year": year,
                    "subject_code": code, "subject_description": desc,
                    "units": float(units), "teacher": teacher, "grade": float(grade)
                })
        if records:
            return pd.DataFrame(records)

    # Fallback: conventional flat table with one row per subject.
    flat = raw.dropna(how="all").copy().reset_index(drop=True)
    if flat.empty:
        raise ValueError("The uploaded file is empty.")
    flat.columns = [clean_text(v) for v in flat.iloc[0].tolist()]
    flat = flat.iloc[1:].reset_index(drop=True)
    name_col = find_column(flat.columns, ["Student Name", "Name", "Full Name", "Student"])
    course_col = find_column(flat.columns, ["Course", "Program", "Degree"])
    year_col = find_column(flat.columns, ["Year", "Year Level", "Level"])
    code_col = find_column(flat.columns, ["Subject Code", "Subject", "Code"])
    desc_col = find_column(flat.columns, ["Subject Description", "Description"])
    units_col = find_column(flat.columns, ["Units", "Unit", "Credit Units"])
    teacher_col = find_column(flat.columns, ["Teacher", "Instructor", "Professor", "Faculty"])
    grade_col = find_column(flat.columns, ["Grade", "Final Grade", "Rating"])
    required = {"Student Name": name_col, "Course": course_col, "Year": year_col,
                "Subject Code": code_col, "Units": units_col, "Teacher": teacher_col,
                "Grade": grade_col}
    missing = [label for label, col in required.items() if col is None]
    if missing:
        raise ValueError("Missing required columns: " + ", ".join(missing))
    result = pd.DataFrame({
        "student_name": flat[name_col].map(clean_text),
        "course": flat[course_col].map(clean_text),
        "year": flat[year_col].map(clean_text),
        "subject_code": flat[code_col].map(clean_text),
        "subject_description": flat[desc_col].map(clean_text) if desc_col else "",
        "units": pd.to_numeric(flat[units_col], errors="coerce"),
        "teacher": flat[teacher_col].map(clean_text),
        "grade": pd.to_numeric(flat[grade_col], errors="coerce")
    })
    return result.dropna(subset=["units", "grade"]).query("units > 0")

# -------------------- GUI --------------------
logo_uri = "data:image/png;base64," + LOGO_B64
header = widgets.HTML(f"""
<div style="font-family:Arial,sans-serif;background:linear-gradient(120deg,{DARK_GREEN},{GREEN});
 color:white;padding:20px 24px;border-radius:16px;display:flex;align-items:center;gap:20px;
 box-shadow:0 6px 18px #0b2f241c;">
 <div style="background:#fff;border:3px solid #fff;border-radius:15px;width:112px;height:112px;
 display:flex;align-items:center;justify-content:center;flex:0 0 112px;overflow:hidden;">
  <img src="{logo_uri}" alt="NDMU seal" style="width:100%;height:100%;object-fit:contain;display:block;">
 </div>
 <div style="min-width:0">
  <div style="font-size:12px;letter-spacing:2px;text-transform:uppercase;color:{GOLD};font-weight:700;">
   Notre Dame of Marbel University</div>
  <h1 style="margin:8px 0 7px;color:#fff;font-size:clamp(22px,3vw,30px);line-height:1.2;">
   Student Grade Report System</h1>
  <div style="font-size:14px;color:#e8f1eb;">Search students · Individual reports · Weighted GWA · Top 10</div>
 </div>
</div>
""")

upload = widgets.FileUpload(accept=".xlsx,.xls,.csv", multiple=False,
                            description="Upload Excel / CSV", button_style="success", icon="upload")
status = widgets.HTML()
search_box = widgets.Text(
    placeholder="Type a student name, course, or year…",
    description="Search:",
    continuous_update=True,
    layout=widgets.Layout(width="100%", max_width="900px")
)
student_list_out = widgets.Output()
report_out = widgets.Output()
leaderboard_out = widgets.Output()
subjects = pd.DataFrame()
summary = pd.DataFrame()
selected_student_id = None

def show_report(student_id):
    global selected_student_id
    if subjects.empty or student_id is None:
        return
    selected_student_id = student_id
    selected = subjects[subjects["_student_id"] == student_id].copy()
    student = summary[summary["_student_id"] == student_id].iloc[0]
    name, course, year = map(lambda x: escape(str(x)),
                             [student["student_name"], student["course"], student["year"]])
    rank = int(student["Rank"])
    with report_out:
        clear_output(wait=True)
        display(HTML(f"""
        <section style="font-family:Arial,sans-serif;border:1px solid #dce7df;border-radius:16px;
        overflow:hidden;margin:8px 0 18px;background:#fff;box-shadow:0 5px 18px #123c320c;">
          <header style="background:linear-gradient(110deg,{DARK_GREEN},{GREEN});padding:20px 22px;color:white;">
            <div style="font-size:10px;letter-spacing:1.8px;color:{GOLD};font-weight:700;">OFFICIAL ACADEMIC SUMMARY</div>
            <h2 style="margin:7px 0 4px;color:#fff;font-size:24px;">{name}</h2>
            <div style="font-size:13px;color:#e4efe8;">{course} &nbsp; | &nbsp; Year {year}</div>
          </header>
          <div style="display:flex;flex-wrap:wrap;gap:10px;padding:14px 16px;background:#f5f8f5;">
            <div style="flex:1;min-width:145px;background:#fff;border:1px solid #dce7df;border-radius:11px;padding:13px;">
              <div style="font-size:10px;color:#64756b;letter-spacing:.7px;">WEIGHTED GWA</div>
              <div style="font-size:29px;font-weight:800;color:{GREEN};margin-top:5px;">{student['GWA']:.2f}</div>
            </div>
            <div style="flex:1;min-width:125px;background:#fff;border:1px solid #dce7df;border-radius:11px;padding:13px;">
              <div style="font-size:10px;color:#64756b;letter-spacing:.7px;">TOTAL UNITS</div>
              <div style="font-size:25px;font-weight:750;color:#24372c;margin-top:7px;">{student['total_units']:g}</div>
            </div>
            <div style="flex:1;min-width:125px;background:{PALE_GOLD};border:1px solid #f0df9e;border-radius:11px;padding:13px;">
              <div style="font-size:10px;color:#78601a;letter-spacing:.7px;">CLASS RANK</div>
              <div style="font-size:25px;font-weight:750;color:#70540b;margin-top:7px;">#{rank}</div>
            </div>
          </div>
          <div style="padding:17px 18px 4px;">
            <h3 style="color:{DARK_GREEN};font-size:16px;margin:0 0 10px;">Subject Grades</h3>
          </div>
        </section>
        """))
        table = selected[["subject_code", "subject_description", "teacher", "units", "grade"]].copy()
        table.columns = ["Subject Code", "Subject Description", "Teacher", "Units", "Grade"]
        display(table.style.hide(axis="index")
            .format({"Units": "{:g}", "Grade": "{:.2f}"})
            .set_table_styles([
                {"selector":"table","props":[("width","100%"),("border-collapse","collapse"),
                    ("font-family","Arial,sans-serif"),("font-size","13px")]},
                {"selector":"th","props":[("background-color",GREEN),("color","white"),
                    ("padding","11px 10px"),("text-align","left")]},
                {"selector":"td","props":[("padding","10px"),("border-bottom","1px solid #e5ece6"),
                    ("color","#26372d")]},
                {"selector":"tbody tr:nth-child(even)","props":[("background-color","#f5f8f5")]}
            ]))
        display(HTML('<p style="font:12px Arial,sans-serif;color:#66756b;margin:12px 2px 20px;">'
                     'Weighted GWA = sum of (grade × units) ÷ total units.</p>'))

def choose_student(button):
    show_report(button.tooltip)

def render_student_list(*args):
    with student_list_out:
        clear_output(wait=True)
        if summary.empty:
            display(HTML('<div style="padding:14px;color:#64756b;">Upload a grade file to see students here.</div>'))
            return
        q = norm(search_box.value)
        matches = summary.copy()
        if q:
            mask = (
                matches["student_name"].map(norm).str.contains(re.escape(q), na=False) |
                matches["course"].map(norm).str.contains(re.escape(q), na=False) |
                matches["year"].map(norm).str.contains(re.escape(q), na=False)
            )
            matches = matches[mask]
        display(HTML(f'<div style="font:12px Arial,sans-serif;color:#64756b;padding:4px 2px 9px;">'
                     f'{len(matches)} student(s) found</div>'))
        if matches.empty:
            display(HTML('<div style="padding:16px;border:1px solid #e5e7eb;border-radius:10px;'
                         'color:#64748b;">No students found. Check the spelling or try a shorter search.</div>'))
            return
        rows = []
        for _, r in matches.iterrows():
            btn = widgets.Button(
                description=f"{r['student_name']}  ·  {r['course']}  ·  Year {r['year']}",
                tooltip=r["_student_id"], icon="user", button_style="",
                layout=widgets.Layout(width="100%", height="42px")
            )
            btn.style.button_color = "#ffffff"
            btn.style.text_color = GREEN
            btn.on_click(choose_student)
            rows.append(btn)
        display(widgets.VBox(rows, layout=widgets.Layout(width="100%", max_width="900px",
                                                          gap="5px")))

def render_leaderboard():
    with leaderboard_out:
        clear_output(wait=True)
        if summary.empty:
            return
        top = summary.head(10)[["Rank", "student_name", "course", "year", "GWA"]].copy()
        top.columns = ["Rank", "Student Name", "Course", "Year Level", "GWA"]
        top["GWA"] = top["GWA"].map(lambda x: f"{x:.2f}")
        display(HTML(f'<h3 style="font-family:Arial,sans-serif;color:{GREEN};margin-bottom:4px;">'
                     'Top 10 Student Leaderboard</h3><p style="font:12px Arial,sans-serif;color:#64756b;margin-top:0;">'
                     'Ranked from lowest weighted GWA to highest.</p>'))
        display(top.style.hide(axis="index").set_table_styles([
            {"selector":"table","props":[("width","100%"),("border-collapse","collapse"),
                ("font-family","Arial,sans-serif"),("font-size","13px")]},
            {"selector":"th","props":[("background-color",GREEN),("color","white"),("padding","11px 10px"),
                ("text-align","left")]},
            {"selector":"td","props":[("padding","10px"),("border-bottom","1px solid #e5ece6")]},
            {"selector":"tbody tr:nth-child(even)","props":[("background-color","#f5f8f5")]}
        ]))

def process_upload(change):
    global subjects, summary, selected_student_id
    if not upload.value:
        return
    try:
        uploaded = upload.value
        if isinstance(uploaded, dict):  # ipywidgets 7
            file_info = next(iter(uploaded.values()))
            content = file_info["content"]
            filename = file_info.get("metadata", {}).get("name", "uploaded_file.xlsx")
        else:  # ipywidgets 8
            file_info = list(uploaded)[0]
            content = file_info["content"]
            filename = file_info.get("name", "uploaded_file.xlsx")
        raw_bytes = io.BytesIO(bytes(content))
        if filename.lower().endswith(".csv"):
            raw = pd.read_csv(raw_bytes, header=None)
        elif filename.lower().endswith((".xlsx", ".xls")):
            raw = pd.read_excel(raw_bytes, header=None)
        else:
            raise ValueError("Please upload a .csv, .xlsx, or .xls file.")

        parsed = parse_repeated_sections(raw)
        if parsed.empty:
            raise ValueError("No valid subject-grade rows were found.")
        for col in ["student_name", "course", "year", "subject_code", "subject_description", "teacher"]:
            parsed[col] = parsed[col].map(clean_text)
        parsed["units"] = pd.to_numeric(parsed["units"], errors="coerce")
        parsed["grade"] = pd.to_numeric(parsed["grade"], errors="coerce")
        parsed = parsed[
            (parsed["student_name"] != "") & (parsed["course"] != "") &
            (parsed["subject_code"] != "") & parsed["units"].notna() &
            parsed["grade"].notna() & (parsed["units"] > 0)
        ].copy()
        if parsed.empty:
            raise ValueError("No complete records remained after validating student names, units, and grades.")

        # Stable identity prevents students with same name but different course/year from merging.
        parsed["_student_id"] = (
            parsed["student_name"].str.casefold() + "||" +
            parsed["course"].str.casefold() + "||" + parsed["year"].str.casefold()
        )
        parsed["_weighted"] = parsed["grade"] * parsed["units"]
        summary_df = parsed.groupby("_student_id", as_index=False).agg(
            student_name=("student_name", "first"), course=("course", "first"),
            year=("year", "first"), total_units=("units", "sum"),
            weighted_total=("_weighted", "sum"), subject_count=("subject_code", "count")
        )
        summary_df["GWA"] = summary_df["weighted_total"] / summary_df["total_units"]
        summary_df = summary_df.sort_values(["GWA", "student_name"], ascending=[True, True]).reset_index(drop=True)
        summary_df["Rank"] = range(1, len(summary_df) + 1)
        subjects = parsed
        summary = summary_df
        selected_student_id = None
        status.value = (f'<div style="font-family:Arial,sans-serif;background:#e8f5eb;color:#185c32;'
                        f'padding:12px 14px;border-radius:9px;margin:12px 0;">'
                        f'<b>Import successful</b> · {escape(filename)}<br>'
                        f'{len(summary)} students · {len(subjects)} subject records</div>')
        render_student_list()
        render_leaderboard()
        if len(summary):
            show_report(summary.iloc[0]["_student_id"])
    except Exception as exc:
        status.value = (f'<div style="font-family:Arial,sans-serif;background:#fde8e7;color:#922b21;'
                        f'padding:12px 14px;border-radius:9px;margin:12px 0;">'
                        f'<b>Could not process this file.</b><br>{escape(str(exc))}</div>')

upload.observe(process_upload, names="value")
search_box.observe(render_student_list, names="value")

display(header)
display(HTML(f'<div style="font-family:Arial,sans-serif;margin:20px 2px 8px;"><h3 style="color:{DARK_GREEN};margin:0 0 5px;">Import student data</h3><p style="color:#64756b;margin:0;">Upload an Excel or CSV file. Your original file will not be modified.</p></div>'))
display(upload)
display(status)
display(HTML(f'<div style="font-family:Arial,sans-serif;margin:22px 2px 8px;"><h3 style="color:{DARK_GREEN};margin:0 0 8px;">Find a student</h3></div>'))
display(search_box)
display(student_list_out)
display(HTML(f'<div style="font-family:Arial,sans-serif;margin:22px 2px 8px;"><h3 style="color:{DARK_GREEN};margin:0 0 8px;">Individual grade report</h3></div>'))
display(report_out)
display(HTML(f'<div style="font-family:Arial,sans-serif;margin:22px 2px 8px;"><h3 style="color:{DARK_GREEN};margin:0 0 8px;">Academic leaderboard</h3></div>'))
display(leaderboard_out)
```
